```python
import os
import subprocess
import json
import csv

# On adapte les chemins car le notebook est dans le dossier 'notebooks/'
# '..' signifie "remonter d'un dossier" pour revenir à la racine de paraspeech
DOSSIER_RAW = "../data/raw"
DOSSIER_AUDIO = "../data/processed/audio/extrait"
FICHIER_REGISTRE = "../data/processed/videos.csv"

SRC = "../src/"

# Créer les dossiers de test s'ils n'existent pas
os.makedirs(DOSSIER_AUDIO, exist_ok=True)
print("Dossiers prêts !")
```

```python
# Modifie le nom si ta vidéo s'appelle autrement pour le test
nom_video_test = "V001.mp4" 
chemin_complet = os.path.join(DOSSIER_RAW, nom_video_test)

commande_probe = [
    'ffprobe', '-v', 'quiet', 
    '-print_format', 'json', 
    '-show_format', '-show_streams', 
    chemin_complet
]

resultat = subprocess.run(commande_probe, capture_output=True, text=True)
infos = json.loads(resultat.stdout)

# On extrait la durée pour vérifier que ça marche
duree = infos.get('format', {}).get('duration')
print(f"Succès ! La vidéo dure {duree} secondes.")
```

```python
chemin_sortie_audio = os.path.join(DOSSIER_AUDIO, "V001.wav")

commande_ffmpeg = [
    'ffmpeg', '-y', 
    '-i', chemin_complet,
    '-vn', 
    '-acodec', 'pcm_s16le', 
    '-ar', '16000', 
    chemin_sortie_audio
]

print("Extraction en cours...")
subprocess.run(commande_ffmpeg, capture_output=True)
print(f"Vérifie ton dossier : {DOSSIER_AUDIO}. Le fichier V001.wav doit y être !")
```
